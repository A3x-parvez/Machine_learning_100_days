```python
import pandas as pd
df = pd.read_csv('placement.csv')
```

# ✅ 1. Using scipy.stats.zscore (Most Common)

```python
from scipy import stats
import numpy as np
import pandas as pd

# Calculate Z-scores
df['z_score'] = np.abs(stats.zscore(df['cgpa']))

# Detect outliers (|z| > 3)
outliers = df[df['z_score'] > 3]
print(outliers)
```

# ✅ 2. Using statsmodels.robust.scale.mad (for Modified Z-score)

For data that is not normally distributed, use the Modified Z-score based on Median Absolute Deviation (MAD):

```python
from statsmodels.robust.scale import mad

# Calculate modified Z-score manually
median = df['cgpa'].median()
mad_val = mad(df['cgpa'])
df['mod_z_score'] = 0.6745 * (df['cgpa'] - median) / mad_val

# Detect outliers
df[np.abs(df['mod_z_score']) > 3.5]
```

# ✅ 3. Using sklearn.preprocessing.StandardScaler (for scaling, not directly outlier detection)

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df['z_scaled'] = np.abs(scaler.fit_transform(df[['cgpa']]))
df[df['z_scaled'] > 3]
```
